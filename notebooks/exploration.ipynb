## SQL Analysis - Business Questions
### Question 1: Is there a relationship between price and rating? (Are expensive products better rated?)

```python
conn = sqlite3.connect('../data/cosmetics.db')
result = pd.read_sql('''
		SELECT 
			CASE
				WHEN Price_USD < 50 THEN '1-Low ($10-$50)'
				WHEN Price_USD < 100 THEN '2-Mid ($50-$100)'
				ELSE '3-High ($100-$150)'
			END AS Price_Range,
			ROUND(AVG(Rating), 2) AS Avg_Rating,
			COUNT(*) AS Num_Products
		FROM products
		GROUP BY Price_Range
		ORDER BY Price_Range ASC
''', conn)
conn.close()
result
```

### Results: Price vs Rating
No significant relationship found between price and rating.
ALL price ranges average 3.0 stars, suggesting that higher price does not guarantee better costumer satisfaction.

### Question 2: Which categories (skincare, makeup, etc.) have the best price/quality ratio?

```python
conn = sqlite3.connect('../data/cosmetics.db')
result = pd.read_sql('''
		SELECT Category,
		ROUND(AVG(Rating), 2) AS Avg_Rating,
		ROUND(AVG(Price_USD), 2) AS Avg_Price,
		ROUND(AVG(Rating)/AVG(Price_USD), 4) AS Value_Index
		FROM products
		GROUP BY Category
		ORDER BY Value_Index DESC
''', conn)
conn.close()
result
```

### Results: Best Price:Quality Ratio by Category
Contour, BB Cream and CC Cream show the best value index.
Lip products (Lipstick, Lip Liner) and Eye Shadow show the lowest price/quality ratio.
Overall differences are small, suggesting pricing is uniform across categories in this market.

### Question 3: How does price vary by skin type or gender target?

```python
conn = sqlite3.connect('../data/cosmetics.db')
result = pd.read_sql('''
		SELECT Skin_Type,
		ROUND(AVG(Price_USD), 2) AS Avg_Price
		FROM products
		GROUP BY SKin_Type
		ORDER BY Avg_price DESC
''', conn)
conn.close()
result
```

```python
conn = sqlite3.connect('../data/cosmetics.db')
result = pd.read_sql('''
		SELECT Gender_Target,
		ROUND(AVG(Price_USD), 2) AS Avg_Price
		FROM products
		GROUP BY Gender_Target
		ORDER BY Avg_price DESC
''', conn)
conn.close()
result
```

### Results: Price by Skin Type and Gender
- Dry skin products are slightly more expensive ($81.51 avg).
- Male-targeted products marginally more expensive than female ($80.26 vs $79.99).
- Differences are minimal (<$2), suggesting uniform pricing strategy regardless of skin type or gender target.

### Question 4:  What percentage of products are cruelty-free, and does it vary by brand or country?

```python
conn = sqlite3.connect('../data/cosmetics.db')
result = pd.read_sql('''
		SELECT Country_of_Origin,
		ROUND(SUM(Cruelty_Free) * 100.0 / COUNT(*), 2) AS Avg_Cruelty_Free
		FROM products
		GROUP BY Country_of_Origin
		ORDER BY Avg_Cruelty_Free DESC
''', conn)
conn.close()
result
```

```python
conn = sqlite3.connect('../data/cosmetics.db')
result = pd.read_sql('''
		SELECT Brand,
		ROUND(SUM(Cruelty_Free) * 100.0 / COUNT(*), 2) AS Avg_Cruelty_Free
		FROM products
		GROUP BY Brand
		ORDER BY Avg_Cruelty_Free DESC
''', conn)
conn.close()
result
```

### Results: Cruelty-Free by Country and Brand

**By Country:**
France leads with 51.58% cruelty-free products, South Korea lowest at 48.19%.
Minimal variation across countries (~3% difference).

**By Brand**
ColourPop leads with 56.1% cruelty-free products.
Make Up For Ever and Tatcha show lowest cruelty-free rates (~43%).
Overall, all brands hover around 50%, no strong cruelty-free leaders emerge.

### Question 5: Which countries of origin have the most presence and how do they compare in rating/price?

```python
conn = sqlite3.connect('../data/cosmetics.db')
result = pd.read_sql('''
		SELECT Country_of_Origin,
		ROUND(AVG(Rating), 2) AS Avg_Rating,
		ROUND(AVG(Price_USD), 2) AS Avg_Price,
		COUNT(*) AS Presence
		FROM products
		GROUP BY Country_of_Origin
		ORDER BY Presence DESC
''', conn)
conn.close()
result
```

### Results: Countries by Presence, Rating and Price
- Italy and USA have the highest product count (~1,942 and 1,931).
- Japan has the highest average rating (3.04) and lowest price ($78.77).
- South Korea has the highest average price ($82.40).
- Overall distribution is very uniform across all countries.

## General Conclusion

The analysis reveals uniformly similar results across all business questions:
- Price does not correlate with rating (~3.0 average across all price ranges).
- Category, skin type, gender, and country show minimal price/rating variation.
- Cruelty-Free rates hover around 50% for all brands and countries.

These uniform results are consistent with q synthetically generated dataset where ratings apper randomly distributed (perfect mean of 3.0).
Real market data would show stronger patterns and outliers.

## Export the data to Tablaeu

```python
import sqlite3
import pandas as pd
import os

os.makedirs('../data/tableau', exist_ok=True)
conn = sqlite3.connect('../data/cosmetics.db')

# Q1 - Price vs Rating
q1 = pd.read_sql('''
    SELECT 
        CASE
            WHEN Price_USD < 50 THEN '1-Low ($10-$50)'
            WHEN Price_USD < 100 THEN '2-Mid ($50-$100)'
            ELSE '3-High ($100-$150)'
        END AS Price_Range,
        ROUND(AVG(Rating), 2) AS Avg_Rating,
        COUNT(*) AS Num_Products
    FROM products
    GROUP BY Price_Range
    ORDER BY Price_Range DESC
''', conn)
q1.to_csv('../data/tableau/q1_price_vs_rating.csv', index=False)

# Q2 - Category Value Index
q2 = pd.read_sql('''
    SELECT Category,
        ROUND(AVG(Rating), 2) AS Avg_Rating,
        ROUND(AVG(Price_USD), 2) AS Avg_Price,
        ROUND(AVG(Rating)/AVG(Price_USD), 4) AS Value_Index
    FROM products
    GROUP BY Category
    ORDER BY Value_Index DESC
''', conn)
q2.to_csv('../data/tableau/q2_category_value.csv', index=False)

# Q3 - Price by Skin Type and Gender
q3_skin = pd.read_sql('''
    SELECT Skin_Type,
        ROUND(AVG(Price_USD), 2) AS Avg_Price
    FROM products
    GROUP BY Skin_Type
    ORDER BY Avg_Price DESC
''', conn)
q3_skin.to_csv('../data/tableau/q3_skin_type.csv', index=False)

q3_gender = pd.read_sql('''
    SELECT Gender_Target,
        ROUND(AVG(Price_USD), 2) AS Avg_Price
    FROM products
    GROUP BY Gender_Target
    ORDER BY Avg_Price DESC
''', conn)
q3_gender.to_csv('../data/tableau/q3_gender.csv', index=False)

# Q4 - Cruelty Free
q4_country = pd.read_sql('''
    SELECT Country_of_Origin,
        ROUND(SUM(Cruelty_Free) * 100.0 / COUNT(*), 2) AS Cruelty_Free_Pct
    FROM products
    GROUP BY Country_of_Origin
    ORDER BY Cruelty_Free_Pct DESC
''', conn)
q4_country.to_csv('../data/tableau/q4_cruelty_country.csv', index=False)

q4_brand = pd.read_sql('''
    SELECT Brand,
        ROUND(SUM(Cruelty_Free) * 100.0 / COUNT(*), 2) AS Cruelty_Free_Pct
    FROM products
    GROUP BY Brand
    ORDER BY Cruelty_Free_Pct DESC
''', conn)
q4_brand.to_csv('../data/tableau/q4_cruelty_brand.csv', index=False)

# Q5 - Countries Analysis
q5 = pd.read_sql('''
    SELECT Country_of_Origin,
        ROUND(AVG(Rating), 2) AS Avg_Rating,
        ROUND(AVG(Price_USD), 2) AS Avg_Price,
        COUNT(*) AS Num_Products
    FROM products
    GROUP BY Country_of_Origin
''', conn)
q5.to_csv('../data/tableau/q5_countries.csv', index=False)

conn.close()
print("All CSVs exported!")
```
